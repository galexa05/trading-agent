```python
import pandas as pd
# Load the articles CSV
articles_path = '../../data/stock_portfolios/articles.csv'
df = pd.read_csv(articles_path)

# Drop records with null summaries
df_cleaned = df.dropna(subset=['summary'])

# Save the cleaned dataframe to CSV
df_cleaned.to_csv('../../data/stock_portfolios/articles_cleaned.csv', index=False)

# Print some stats
print(f"Original dataset: {len(df)} records")
print(f"Cleaned dataset: {len(df_cleaned)} records")
print(f"Removed {len(df) - len(df_cleaned)} records with null summaries")
```
